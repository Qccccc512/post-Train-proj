# PyTorch Low-Load Keep-Alive Loop

Run the code cell below to execute a tiny matrix multiplication every 1 second.
It is designed to keep the notebook active with minimal resource usage.
By default, it tries to run on GPU first and falls back to CPU.

Stop by interrupting the cell (Runtime -> Interrupt execution).

```python
import time
import torch

# Set to False if you do not want to occupy GPU at all.
USE_GPU = True

if USE_GPU and torch.cuda.is_available():
    device = "cuda"
else:
    device = "cpu"
    # Reduce host load when running on CPU.
    torch.set_num_threads(1)

a = torch.tensor([[1.0, 2.0], [3.0, 4.0]], device=device)
b = torch.tensor([[5.0, 6.0], [7.0, 8.0]], device=device)

print(f"Keep-alive loop started on {device}. Interrupt this cell to stop.")
counter = 0

with torch.no_grad():
    while True:
        _ = a @ b
        counter += 1

        # Print once per minute to avoid large notebook outputs.
        if counter % 60 == 0:
            print(f"alive: {counter} iterations")

        # Simple mode: sleep 1 second after each multiplication.
        time.sleep(1.0)
```
